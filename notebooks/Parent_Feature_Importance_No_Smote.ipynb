```python
import pandas as pd
import os
import matplotlib.pyplot as plt
from hbn.constants import Defaults
```

```python
# load model data
MODEL_DIR = os.path.join(Defaults.MODEL_DIR, '2023-03-10_01-47-37-37S/model_30732935') 

model_auc = os.path.join(MODEL_DIR, plt.imread("out-localspec-20230310T093539.656953/test-roc_auc_score-20230310T093542.637069.png"))
df_feat = pd.read_csv(os.path.join(MODEL_DIR, 'classifier-feature_importance.csv'))
df_perm = pd.read_csv(os.pth.join(MODEL_DIR, 'classifier-permutation_importance.csv'))
```

```python
plt.imshow(model_auc)
```

```python
# splitting up dataframes by classifier
df_decision_feat = df_feat[df_feat['clf'] == "DecisionTreeClassifier"]
df_logistic_feat = df_feat[df_feat['clf'] == "LogisticRegressionCV"]

df_decision_perm = df_perm[df_perm['clf'] == "DecisionTreeClassifier"]
df_logistic_perm = df_perm[df_perm['clf'] == "LogisticRegressionCV"]
```

```python
df_decision_feat.nlargest(10, 'feature_probabilities_rank_order')
```

Top Features by Feature Importance for Decision Tree

```python
df_decision_feat.nlargest(10, 'feature_probabilities_rank_order').plot.barh(x = 'feature_names_rank_order', y = 'feature_probabilities_rank_order')
```

Top Features by Permutation Importance for Decision Tree

```python
df_decision_perm.nlargest(20, 'feature_probabilities_rank_order').plot.barh(x = 'feature_names_rank_order', y = 'feature_probabilities_rank_order')
```
